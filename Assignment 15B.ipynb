```python
#Q2
df.groupby("gender")["cups_per_day"].mean().round(1)

```

```python
#Q3
counts = df['snack_choice'].value_counts()

ax = counts.plot(kind='bar', title='Snack Choices of Students', rot=0)
ax.set_xlabel('Snack')
ax.set_ylabel('Number of students')
ax.figure.savefig('snack_choices.png')
```
